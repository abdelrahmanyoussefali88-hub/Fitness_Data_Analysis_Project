```python
import pandas as pd

# إنشاء بيانات العملاء (Dataset)
data = {
    'Name': ['Khaled', 'Sara', 'Abdelrahman', 'Nour', 'Tarek'],
    'Age': [25, 22, 29, 27, 34],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Height_cm': [175, 162, 178, 165, 185],
    'Weight_kg': [75, 60, 80, 68, 95],
    'Activity_Level': ['Moderate', 'Light', 'High', 'Moderate', 'Very High'],
    'Goal': ['Maintenance', 'Weight Loss', 'HIT Training', 'Weight Loss', 'Bulking']
}

# تحويل البيانات لـ DataFrame عشان نقدر نحللها
df = pd.DataFrame(data)

# عرض البيانات للتأكد إنها اتحملت صح
df
```

```python
# تحويل الطول من سنتيمتر لمتر
height_m = df['Height_cm'] / 100

# حساب الـ BMI وإضافته في عمود جديد في الجدول
df['BMI'] = df['Weight_kg'] / (height_m ** 2)

# تقريب النتيجة لعلامتين عشريتين عشان الجدول يكون منظم
df['BMI'] = df['BMI'].round(2)

# عرض الجدول بعد إضافة عمود الـ BMI
df
```

```python
# إنشاء دالة لحساب الـ BMR
def calculate_bmr(row):
    weight = row['Weight_kg']
    height = row['Height_cm']
    age = row['Age']
    gender = row['Gender']

    if gender == 'Male':
        bmr = (10 * weight) + (6.25 * height) - (5 * age) + 5
    else:
        bmr = (10 * weight) + (6.25 * height) - (5 * age) - 161

    return bmr

# تطبيق الدالة على الجدول وإضافة عمود جديد
df['BMR'] = df.apply(calculate_bmr, axis=1)

# تقريب الرقم لعدد صحيح
df['BMR'] = df['BMR'].round(0).astype(int)

# عرض الجدول
df
```

```python
# دالة لحساب السعرات النهائية وتوزيع الماكروز
def calculate_tdee_and_macros(row):
    bmr = row['BMR']
    activity = row['Activity_Level']
    goal = row['Goal']
    weight = row['Weight_kg']

    # 1. تحديد معامل النشاط لحساب إجمالي الحرق (TDEE)
    if activity == 'Light':
        tdee = bmr * 1.375
    elif activity == 'Moderate':
        tdee = bmr * 1.55
    elif activity == 'High':
        tdee = bmr * 1.725
    else: # Very High
        tdee = bmr * 1.9

    # 2. تعديل السعرات حسب الهدف الرياضي
    if goal == 'Weight Loss':
        target_calories = tdee - 500
    elif goal == 'Bulking':
        target_calories = tdee + 500
    else: # Maintenance أو HIT Training (المحافظة على الوزن مع تغيير التكوين العضلي)
        target_calories = tdee

    # 3. توزيع الماكروز
    # البروتين (2.2 جرام لكل كيلو من الوزن) - الجرام فيه 4 سعرات
    protein_grams = weight * 2.2
    protein_calories = protein_grams * 4

    # الدهون (25% من إجمالي السعرات) - الجرام فيه 9 سعرات
    fat_calories = target_calories * 0.25
    fat_grams = fat_calories / 9

    # الكارب (باقي السعرات) - الجرام فيه 4 سعرات
    carb_calories = target_calories - (protein_calories + fat_calories)
    carb_grams = carb_calories / 4

    # إرجاع القيم كأرقام صحيحة
    return pd.Series([int(target_calories), int(protein_grams), int(carb_grams), int(fat_grams)])

# تطبيق الدالة لإنشاء 4 أعمدة جديدة في الجدول
df[['Target_Calories', 'Protein_g', 'Carbs_g', 'Fats_g']] = df.apply(calculate_tdee_and_macros, axis=1)

# عرض الجدول النهائي
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# تحديد حجم وشكل الرسم البياني
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# إنشاء رسم بياني للأعمدة (Bar Chart)
ax = sns.barplot(x='Name', y='Target_Calories', hue='Goal', data=df, dodge=False, palette='Set2')

# إضافة العناوين للرسم
plt.title('Target Calories Based on Fitness Goals', fontsize=16, weight='bold')
plt.xlabel('Name', fontsize=12)
plt.ylabel('Target Calories (kcal)', fontsize=12)

# وضع مفتاح الألوان (Legend) في مكان مناسب
plt.legend(title='Fitness Goal', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# عرض الرسم البياني
plt.show()
```
